```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
arrays = []
hr_files = os.listdir("./hycom_numpy")

for file in hr_files:
    arr = np.load("./hycom_numpy/" + file).astype(np.float32)

    print(f"{file}: {arr.shape}")

    print(f"  NaNs : {np.isnan(arr).sum()}")
    print(f"  Infs : {np.isinf(arr).sum()}")
    print(f"  Min  : {np.nanmin(arr)}")
    print(f"  Max  : {np.nanmax(arr)}")

    arrays.append(arr)

hr_data = np.concatenate(arrays, axis=0)

print("\nWhole dataset")
print(f"Shape : {hr_data.shape}")
print(f"NaNs  : {np.isnan(hr_data).sum()}")
print(f"Infs  : {np.isinf(hr_data).sum()}")
print(f"Min   : {np.nanmin(hr_data)}")
print(f"Max   : {np.nanmax(hr_data)}")
```
